### MODIFICACIÓN Y LIMPIEZA DE LOS DATOS

```python
# Importamos las librerías necesarias:

import pandas as pd
import numpy as np

import re

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
# Cragamos nuestra base de datos:

df0 = pd.read_csv("HR RAW DATA.csv", index_col=0)
df0
```

```python
# Cargamos el csv creado con una nueva columna department sin datos nulos. Rellenada en base a los datos existentes de la columna jobrole y department. De esta manera también
#podríamos usar esta variable para el análisis (este paso esta hecho en otro Jupyter):

df1 = pd.read_csv('Columna_department.csv', index_col = 0)
```

```python
# Renombramos la columna department de nuestra base de datos inicial, así podremos diferenciarla de la columna nueva creada.

df0.rename(columns={'Department':'Department1'}, inplace=True)
```

```python
# Comprobamos que realmente se ha cambiado el nombre de la columna:

df0
```

```python
# Unimos los dos DataFrames por columnas, así ya tenemos las 2 variables de department en la misma base de datos:

df = pd.concat([df0, df1], axis=1)
df
```

```python
# Eliminamos las columnas que según el primer análisis exploratorio realizado de los datos no nos interesan y también la columna inicial de department, que tiene muchos valores nulos.

df.drop(['Department1', 'DailyRate', 'YearsInCurrentRole', 'HourlyRate', 'MonthlyRate', 'NUMBERCHILDREN', 'Over18', 'PerformanceRating', 'TOTALWORKINGYEARS', 'SameAsMonthlyIncome', 'RoleDepartament', 'DateBirth', 'Salary', 'employeecount'], axis=1, inplace=True)
```

```python
# Comprobamos que se hayan eliminado las columnas:

df.head(1)
```

```python
# Empezamos a limpiar los datos. Primero, pasamos todos los nombres de las columnas a minúsculas y quitamos los espacios iniciales y/o finales:

nuevas_columnas = {}
for i in df.columns:
    nuevas_columnas[i] = i.lower().strip()

df.rename(columns = nuevas_columnas, inplace=True)
df.sample(1)
```

```python
# Ahora recordamos el tipo de dato que tenemos en cada columna:

df.info()
```

```python
# Mostramos el porcentage de valores nulos que tenemos en cada columna:

round((df.isnull().sum() / df.shape[0]) * 100, 2)
```

```python
# CAMBIAMOS LOS TIPOS DE DATOS DE LAS COLUMNAS NECESARIAS:

# Para convertir los tipos de datos de varias columnas a la vez:

df = df.astype({'gender': str, 'remotework': str})
df.info()
```

```python
# Pasamos los datos de todas las columnas tipo object a minúsculas:

columnas = ["attrition", "businesstravel", "department", "educationfield", "gender", "jobrole", "maritalstatus", "overtime", "standardhours", "remotework"]
for columna in columnas:
    df[columna] = df[columna].str.lower()
df.head(3)
```

```python
# Las columnas que queremos que sean de tipo 'float', les cambiamos la "," por el "."

columnas_float = ['employeenumber', 'monthlyincome', 'worklifebalance']
for columna in columnas_float:
    df[columna] = df[columna].str.replace(",", ".")

df
```

```python
# Ahora cambiamos las columnas que queremos pasar a numéricas

df = df.astype({'employeenumber': float, 'monthlyincome': float, 'worklifebalance': float})

df.info()
```

```python
# Repasamos los valores de la columna "age", ya que en el análisis exploratorio nos dimos cuenta que 
#tenemos edades escritas en letras y no números, así que si no lo cambiamos no podremos pasar a 'int'.

print(df['age'].unique())
```

```python
# Creamos un diccionario con todas las edades que tenemos para cambiar los datos de la columna a número
# y así poder pasar el tipo de dato a 'int':

dicc_age = {'51': '51', '52': '52', '42': '42', '47': '47', '46': '46', '48': '48', '59': '59', '41': '41', '56': '56', '38': '38', '55': '55',
       '40': '40', '58': '58', '35': '35', '45': '45', '33': '33', '36': '36', '34': '34', 'forty-seven': '47', '53': '53',
       '43': '43', '60': '60', '32': '32', '37': '37', '49': '49', '39': '39', '50': '50', '44': '44', '30': '30',
       'fifty-eight': '58', '29': '29', '31': '31', '54': '54', '57': '57', '27': '27', 'thirty-six': '36', '28': '28',
       '26': '26', 'fifty-five': '55', '25': '25', 'fifty-two': '52', 'thirty-one': '31', '24': '24',
       'thirty': '30', '23': '23', '22': '22', '21': '21', '20': '20', 'twenty-six': '26', '19': '19',
       'thirty-seven': '37', '18': '18', 'thirty-two': '32', 'twenty-four': '24'}

df['age'] = df['age'].map(dicc_age)
print(f"{df['age'].unique()}")
print(f"{df['age'].nunique()}")
```

```python
# Ahora ya podemos cambiar el tipo de dato de 'age':

df['age'] = df['age'].astype(int)
```

```python
# Ahora revisamos los valores de 'maritalstatus', ya que también vimos que 'married' no está bien escrito
# en todos los registros:

df['maritalstatus'].unique()
```

```python
# Cambiamos el 'marreid' por 'married':

df['maritalstatus'].replace('marreid', 'married', inplace=True)
df['maritalstatus'].unique()
```

```python
# En la columna de 'gender', cambiamos el 0 y 1 a m (masculino) y f (femenino), para entenderlo mejor:

diccionario_gender = {'0': "m", '1': "f"} 
df["gender"] = df["gender"].map(diccionario_gender)
df['gender'].unique()
```

```python
# En esta columna nos han comentado que los valores nulos en realidad tendrían que ser de 'jornada completa'
# y los valores que aparece '80', son empleados que estan a jornada parcial. Así que hacemso las modificaciones
# necesarias:

df['standardhours'] = df['standardhours'].fillna('full_time')
df["standardhours"] = df["standardhours"].replace('80,0', 'part_time')
df['standardhours'].unique()
```

```python
# En la columna de 'businesstravel', habíamos visto que las palabras estaban separadas por guión bajo, 
# menos el 'non-travel', que estaba separado por guión. Así que decidimos reemplazarlo:

df['businesstravel'] = df['businesstravel'].replace('non-travel', 'non_travel')
df['businesstravel'].unique()
```

```python
# En la columna de 'distancefromhome', nos comentan que ha habído un error y hay algunos valores que aparecen
# con un guión delante, pero es un error.

print(df['distancefromhome'].unique())
print(df['distancefromhome'].nunique())
```

```python
# Decidimos pasar todos los valores a valor absoluto, y así nos quedamos con los datos sin el negativo
# delante:

df['distancefromhome'] = df['distancefromhome'].abs() #convertir a valor absoluto
print(df['distancefromhome'].unique())
print(df['distancefromhome'].nunique())
```

```python
# También nos han comentado que hay empleados que en la columna de 'environmentsatisfaction' se han equivocado
# y han entrado dos números, pero realmente nos tenemos que quedar con el primero. Así que decidimos
# crear una columna nueva, pasar los datos a tipo object para poder gestionarlo y queddarnos con el valor de la primera
# posición en todos los registros:

df['environment_satisfaction'] = df['environmentsatisfaction'].astype(str).str[0]
df
```

```python
# Posteriormente, eliminamos la columna original y nos quedamos con la modificada, la volvemos a 
# pasar a dato de tipo 'int' y le ponemos el nombre de la columna original.

df.drop('environmentsatisfaction', axis=1, inplace=True)
df['environment_satisfaction'] = df['environment_satisfaction'].astype(int)
df.rename(columns = {'environment_satisfaction': 'environmentsatisfaction'}, inplace= True)
df
```

```python
# Por último, limpiamos los datos de la columna 'remotework'. Ya que cada empleado ha entrado la 
# respuesta a su manera:

df['remotework'].unique()
```

```python
# Creamos un diccionario con las equivalencias a si hacen teletrabajo ('yes') o no ('no'):

dicc_remotework = {'1': 'yes', 'true': 'yes', 'yes': 'yes', '0': 'no', 'false': 'no'}
df['remotework'] = df['remotework'].map(dicc_remotework)

df['remotework'].unique()
```

```python
# Finalmente guardamos el DataFrame limpio a un csv, para poder empezar a tratar los valores nulos:

df.to_csv('HR RAW DATA transformado 02-06.csv', index=False)
```
